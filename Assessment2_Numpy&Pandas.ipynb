**Q10.** Handling Missing Values

Consider the following DataFrame:
df = pd.DataFrame({
'Name': ['A', 'B', 'C', 'D'],
'Age': [21, np.nan, 23, 22],
'Score': [80, 75, np.nan, 90]
})

Write Pandas code to:
Identify the missing values.
Replace the missing Age with the mean age.
Replace the missing Score with the mean score.

```python
df = pd.DataFrame({
    'Name': ['A', 'B', 'C', 'D'],
    'Age': [21, np.nan, 23, 22],
    'Score': [80, 75, np.nan, 90]
})
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Score'] = df['Score'].fillna(df['Score'].mean())
print(df)
```

**Q13.** Feature Creation

Create a new column called:
Average_Technical_Score
Calculate it using:
Python_Score
SQL_Score
Pandas_Score
NumPy_Score
Then create another column called:
Performance
using these rules:
Average Technical Score Performance
>= 90 Excellent
>= 75 Good
>= 60 Average
< 60 Needs Improvement
Display the resulting DataFrame.

```python
technical_cols = ['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']
df['Average_Technical_Score'] = df[technical_cols].mean(axis=1)
def get_performance(score):
    if score >= 90:
        return 'Excellent'
    elif score >= 75:
        return 'Good'
    elif score >= 60:
        return 'Average'
    else:
        return 'Needs Improvement'
df['Performance'] = df['Average_Technical_Score'].apply(get_performance)
display(df[['Name'] + technical_cols + ['Average_Technical_Score', 'Performance']].head(10))
```

**Q14.** GroupBy Analysis


Using the student dataset, calculate the following for each department:
Number of students
Average Final Score
Maximum Final Score
Minimum Final Score
Sort the resulting DataFrame by Average Final Score in descending order.


```python
df['Department'] = df['Department'].str.strip().str.title()
groupby_df = df.groupby('Department')['Final_Score'].agg(
    Number_of_Students='count',
    Average_Final_Score='mean',
    Maximum_Final_Score='max',
    Minimum_Final_Score='min'
).reset_index()
sorted_groupby_df = groupby_df.sort_values(by='Average_Final_Score', ascending=False)
display(sorted_groupby_df)
```
